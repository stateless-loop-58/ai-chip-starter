## FPGA Matrix Multiplication Test

```python
from pynq import Overlay
from pynq import allocate
import numpy as np
```

```python
# Program bitstream to FPGA
overlay = Overlay('/home/ubuntu/workspace/systolic_mmult.bit')
```

```python
# AXI DMA initialization
dma = overlay.axi_dma_0
dma_send = overlay.axi_dma_0.sendchannel
dma_recv = overlay.axi_dma_0.recvchannel
```

```python
# Allocate physical memory
input_buffer = allocate(shape=(8,), dtype=np.uint32)
output_buffer = allocate(shape=(8,), dtype=np.uint64)
```

```python
# Write input B
input_buffer[0] = (( 4 << 24) | ( 3 << 16) | ( 2 << 8) | (1));
input_buffer[1] = (( 8 << 24) | ( 7 << 16) | ( 6 << 8) | (5));
input_buffer[2] = ((12 << 24) | (11 << 16) | (10 << 8) | (9));
input_buffer[3] = ((16 << 24) | (15 << 16) | (14 << 8) | (13));
# Write input A
input_buffer[4] = (( 4 << 24) | ( 3 << 16) | ( 2 << 8) | (1));
input_buffer[5] = (( 8 << 24) | ( 7 << 16) | ( 6 << 8) | (5));
input_buffer[6] = ((12 << 24) | (11 << 16) | (10 << 8) | (9));
input_buffer[7] = ((16 << 24) | (15 << 16) | (14 << 8) | (13));
for i in range(8):
    print("0x%08X" % (input_buffer[i]))
```

```python
# Do AXI DMA MM2S transfer
dma_send.transfer(input_buffer)
# Do AXI DMA S2MM transfer
dma_recv.transfer(output_buffer)
```

```python
# Read output Y
print("%3d " % ((int(output_buffer[0]) & 0x00000000FFFFFFFF) >> 0 ), end="")
print("%3d " % ((int(output_buffer[0]) & 0xFFFFFFFF00000000) >> 32), end="")
print("%3d " % ((int(output_buffer[1]) & 0x00000000FFFFFFFF) >> 0 ), end="")
print("%3d " % ((int(output_buffer[1]) & 0xFFFFFFFF00000000) >> 32))
print("%3d " % ((int(output_buffer[2]) & 0x00000000FFFFFFFF) >> 0 ), end="")
print("%3d " % ((int(output_buffer[2]) & 0xFFFFFFFF00000000) >> 32), end="")
print("%3d " % ((int(output_buffer[3]) & 0x00000000FFFFFFFF) >> 0 ), end="")
print("%3d " % ((int(output_buffer[3]) & 0xFFFFFFFF00000000) >> 32))
print("%3d " % ((int(output_buffer[4]) & 0x00000000FFFFFFFF) >> 0 ), end="")
print("%3d " % ((int(output_buffer[4]) & 0xFFFFFFFF00000000) >> 32), end="")
print("%3d " % ((int(output_buffer[5]) & 0x00000000FFFFFFFF) >> 0 ), end="")
print("%3d " % ((int(output_buffer[5]) & 0xFFFFFFFF00000000) >> 32))
print("%3d " % ((int(output_buffer[6]) & 0x00000000FFFFFFFF) >> 0 ), end="")
print("%3d " % ((int(output_buffer[6]) & 0xFFFFFFFF00000000) >> 32), end="")
print("%3d " % ((int(output_buffer[7]) & 0x00000000FFFFFFFF) >> 0 ), end="")
print("%3d " % ((int(output_buffer[7]) & 0xFFFFFFFF00000000) >> 32))
```

```python
# Delete buffer to prevent memory leak
del input_buffer, output_buffer
```
