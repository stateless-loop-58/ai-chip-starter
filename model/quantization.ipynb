## Quantization Example

```python
import numpy as np
```

```python
# Example FP32 values
x_float = np.array(
    [-2.5, -1.2, -0.5, 0.3, 1.0, 1.8, 2.7],
    dtype=np.float32
)

print("Original FP32:")
print(x_float)
```

```python
# ---------------------------------------------------------
# 1. Calculate quantization scale
# ---------------------------------------------------------

max_value = np.max(np.abs(x_float))

scale = max_value / 127.0 # INT8 range -128 to 127

print("\nQuantization scale:")
print(scale)
```

```python
# ---------------------------------------------------------
# 2. Quantize: FP32 -> INT8
# ---------------------------------------------------------

x_int8 = np.round(x_float / scale)
x_int8 = np.clip(x_int8, -128, 127).astype(np.int8)

print("\nQuantized INT8:")
print(x_int8)
```

```python
# ---------------------------------------------------------
# 3. Dequantize: INT8 -> FP32
# ---------------------------------------------------------

x_dequantized = x_int8.astype(np.float32) * scale

print("\nDequantized FP32:")
print(x_dequantized)
```

```python
# ---------------------------------------------------------
# 4. Calculate quantization error
# ---------------------------------------------------------

error = np.abs(x_float - x_dequantized)

print("\nQuantization error:")
print(error)

print(f"\nMean Absolute Error: {np.mean(error):.6f}")
print(f"Maximum Absolute Error: {np.max(error):.6f}")
```
